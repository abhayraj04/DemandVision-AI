# MarketVision AI: Notebook 03 - Exploratory Data Analysis
## Distribution of Returns, Volatility Clustering, and Cross-Asset Correlation

This notebook conducts financial econometric exploratory analysis:
1. **Stationarity & Autocorrelation**: Testing Augmented Dickey-Fuller (ADF) on prices vs returns.
2. **Stylized Empirical Facts**: Fat tails (leptokurtosis), volatility clustering (Mandelbrot effect).
3. **Cross-Asset Correlation Matrix**: Correlation structure across blue-chip NSE equities.

```python
import os
import sys
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import adfuller

sys.path.append(os.path.abspath('..'))

# Load cleaned asset series
df_rel = pd.read_parquet('../backend/data/processed/RELIANCE.parquet')

# ADF Test on Price vs Log Return
adf_price = adfuller(df_rel['close'])
adf_ret = adfuller(df_rel['log_return'].dropna())

print(f"ADF Statistic (Close Price): {adf_price[0]:.4f}, p-value: {adf_price[1]:.4e} (Non-Stationary)")
print(f"ADF Statistic (Log Returns): {adf_ret[0]:.4f}, p-value: {adf_ret[1]:.4e} (Stationary, I(0))")
```
