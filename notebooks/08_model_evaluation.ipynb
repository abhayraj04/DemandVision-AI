# MarketVision AI: Notebook 08 - Model Evaluation & Residual Diagnostics
## Leaderboard Benchmarking, Error Distribution, and Selection Rationale

In institutional quantitative finance, model complexity must earn its place.
This notebook explores:
1. **Occam's Razor & Overfitting**: Why complex neural networks or ensembles often underperform in high-noise market regimes.
2. **Residual Diagnostics**: Testing $\epsilon_t = y_t - \hat{y}_t$ for zero-mean, skewness, and kurtosis.
3. **Multi-Model Leaderboard**: Side-by-side comparison across all instruments.

```python
import os
import sys
import json
import pandas as pd
import numpy as np

sys.path.append(os.path.abspath('..'))

with open('../backend/data/reports/model_leaderboard.json', 'r') as f:
    board = json.load(f)

summary = []
for sym, val in board.items():
    summary.append({
        'Symbol': sym,
        'Best Model': val['best_model'],
        'Best RMSE': val['models'][val['best_model']]['rmse'],
        'Best Dir Acc (%)': val['models'][val['best_model']]['directional_accuracy'],
        'Selection Rationale': val['selection_rationale'][:60] + '...'
    })

pd.DataFrame(summary)
```
